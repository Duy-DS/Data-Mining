## Tải và khám phá dữ liệu

```python
import pandas as pd
# Or load titanic data that are alraedy split into train and test data sets according to https://www.kaggle.com/c/titanic/data
# But the test data of kaggle does not have labels
# Therefore we will load  the whole data from a data repository then split it latter
titanic_data = pd.read_csv("https://www.openml.org/data/get_csv/16826755/phpMYEkMl.csv", na_values=['?']) #yo
titanic_data.head()
```

**Các giá trị được coi là “missing”**

Có nhiều cách để biểu diễn các giá trị bị thiếu trong cả tệp dữ liệu và python pandas.

Các giá trị bị thiếu trong dữ liệu có thể là các mục trống, hoặc '?', hoặc một ký tự khác mà người thu thập dữ liệu đã thống nhất để biểu diễn dữ liệu chưa được quan sát.
Trong trường hợp này là '?' -- biết được điều này, chúng ta cho `pandas` biết những gì cần được coi là giá trị bị thiếu thông qua `na_values=['?']`.

Ở "đầu kia", `pandas` có thể biểu diễn các giá trị bị thiếu theo nhiều cách khác nhau. Như có thể thấy ở trên, "NaN" là ký hiệu giá trị bị thiếu mặc định, tuy nhiên, chúng ta cần có khả năng dễ dàng phát hiện giá trị này với dữ liệu thuộc các kiểu khác nhau: dấu phẩy động, số nguyên, boolean và đối tượng chung. Tuy nhiên, trong nhiều trường hợp, một số dạng khác có thể tham chiếu đến các giá trị bị thiếu, chẳng hạn như None "thiếu" hoặc "không khả dụng", "NA" hoặc (-)inf.

```python
# Chúng ta hãy bỏ qua một số tính năng mà chúng ta sẽ không xem xét ở đây.
titanic_data.drop(['name','ticket', 'embarked', 'boat' ,'body' ,'home.dest'], axis=1, inplace=True)
```

Bây giờ chúng ta sẽ chia dữ liệu để đào tạo và kiểm tra các tập hợp con vì **CHỈ** dữ liệu đào tạo sẽ được sử dụng để tìm hiểu các yếu tố đầu vào sau đó các mô hình đã học được sẽ được áp dụng cho dữ liệu kiểm tra

```python
from sklearn.model_selection import train_test_split
y = titanic_data['survived']
X = titanic_data.drop(['survived'], axis=1)
X_titamic_train, X_titanic_test, y_titanic_train, y_titanic_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Bây giờ nếu chúng ta thực hiện phân loại thì nó có thể không hiệu quả với hầu hết các bộ phân loại
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier()
#classifier=SVC()
classifier.fit(X_titamic_train, y_titanic_train)
```

# Có một vấn đề là một số tính năng chứa các giá trị chuỗi, cụ thể là các tính năng "sex" và "cabin", vì vậy hãy mã hóa các tính năng này

```python
import numpy as np
# Mã hóa các đặc điểm phân loại bằng cách bảo toàn các giá trị bị thiếu trong các đặc điểm không đầy đủ
from sklearn.preprocessing import OrdinalEncoder
encoder_sex = OrdinalEncoder(handle_unknown= 'use_encoded_value', unknown_value=np.nan)
X_titamic_train_encoded = X_titamic_train.copy()
X_titamic_train_encoded['sex'] = encoder_sex.fit_transform(X_titamic_train_encoded['sex'].values.reshape(-1,1))

# Bây giờ hãy mã hóa hoàn toàn tính nang Cabin
encoder_cabin = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=np.nan) # Bạn có thể sử dụng cùng một bộ mã hóa cho cả hai nhưng chúng tôi sử dụng hai bộ để làm rõ hơn
X_titamic_train_encoded['cabin'] = encoder_cabin.fit_transform(X_titamic_train_encoded['cabin'].values.reshape(-1, 1).astype(str))
# Lấy mã của những giá trị 'nan' cho đặc trưng cabin
cabin_nan_code=encoder_cabin.transform([['nan']])[0][0]
#In (cabin_nan_code)
# Bây giờ, lấy lại các giá trị nan bị thiếu trong dữ liệu được mã hóa
X_titamic_train_encoded['cabin'].replace(cabin_nan_code, np.nan, inplace = True)
```

## `X_titanic_train_encoded` đã được mã hóa hoàn toàn trong dữ liệu huấn luyện

```python
# Kiểm tra loại dữ liệu đã được mã hóa, những đặc trưng khong object
X_titamic_train_encoded.info()
```

```python
# Như vậy dữ liệu đã khong còn strings/object, hãy thử thực hiện phân loại lại
classifier.fit(X_titamic_train_encoded, y_titanic_train)
```

Hơn nữa, phương pháp xóa các trường hợp không đầy đủ không thực tế đối với dữ liệu thử nghiệm: chúng tôi thực sự muốn biết câu trả lời!

## Bù khuyết bằng pandas

### Bù khuyến đơn giản

```python
# Trung bình giá trị dạng số
X_titamic_data_complete = X_titamic_train_encoded.copy()
X_titamic_data_complete['age'] = X_titamic_data_complete['age'].fillna(X_titamic_data_complete['age'].mean())
X_titamic_data_complete['fare'] = X_titamic_data_complete['fare'].fillna(X_titamic_data_complete['fare'].mean())
X_titamic_data_complete['cabin'] = X_titamic_data_complete['cabin'].fillna(X_titamic_data_complete['cabin'].mean())

#Hiển thị số giá trị thiếu
print(X_titamic_data_complete.isnull().sum())
```

```python
X_titamic_data_complete.head()
```

## Bù khuyết bằng nội suy trong pandas

```python
X_titamic_data_complete = X_titamic_train_encoded.copy()
X_titamic_train_complete = X_titamic_train_complete.interpolate()
# Đầu ra là 'numpy.ndarray' vì vậy chúng ta chuyển nó và dataframe để nhất quán
X_titamic_train_complete = pd.DataFrame(X_titamic_train_complete)
print(X_titamic_train_complete.isna().sum())
```

## Bù khuyết bằng sklearn

### Bù khuyết bằng sklearn đơn giản

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer()
X_titamic_train_complete = imputer.fit_transform(X_titamic_train_encoded)
# Đầu ra là 'numpy.ndarray' vì vậy chúng ta chuyển nó và dataframe để nhất quán
X_titamic_train_complete = pd.DataFrame(X_titamic_train_complete, columns=X_titamic_train_encoded.columns)
print('Số giá trị thiếu: \n', X_titamic_train_complete.isnull().sum())
```

## Chiến lược mặc định cho sklearn simple imputer là "mean", bạn có thể thay đổi nó bằng cách sử dụng tham số strategy

```python
imputer = SimpleImputer(strategy='median')
X_titamic_train_complete = imputer.fit_transform(X_titamic_train_encoded)
# Đầu ra là 'numpy.ndarray' vì vậy chúng ta chuyển nó và dataframe để nhất quán
X_titamic_train_complete = pd.DataFrame(X_titamic_train_complete, columns=X_titamic_train_encoded.columns)
print('Số lượng giá trị thiếu: \n', X_titamic_train_complete.isnull().sum())
```

## Bù khuyết với kNN (sklearn)

```python
from sklearn.impute import KNNImputer
imputer = KNNImputer()
X_titamic_train_complete = imputer.fit_transform(X_titamic_train_encoded)
# Đầu ra là 'numpy.ndarray' vì vậy chúng ta chuyển nó và dataframe để nhất quán
X_titamic_data_complete = pd.DataFrame(X_titamic_train_complete, columns= X_titamic_train_encoded.columns)
print("Số lượng giá trị thiếu: \n", X_titamic_data_complete.isnull().sum())
```

## Bù khuyết lặp (sklearn)

Lưu ý đây là cách sklearn triển khai phương pháp ban đầu được gọi là "MICE" -- hãy xem bài giảng số 2 trong tuần này để biết thêm giải thích.

```python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
imputer = IterativeImputer()
X_titamic_train_complete = imputer.fit_transform(X_titamic_train_encoded)
# Đầu ra là 'numpy.ndarray' vì vậy chúng ta chuyển nó và dataframe để nhất quán
X_titamic_train_complete = pd.DataFrame(X_titamic_train_complete, columns = X_titamic_train_encoded.columns)
print("Số lượng giá trị thiếu: \n", X_titamic_train_complete.isnull().sum())
```

Bạn có thể đặt lại các tham số mặc định của bộ nhập lặp. Ví dụ: bạn có thể đặt số lần lặp. Hơn nữa, bạn có thể chỉ định bộ ước lượng để ước lượng các giá trị còn thiếu.

```python
# Hãy sử dung DT để ước lượng
from sklearn.tree import DecisionTreeRegressor
imputer = IterativeImputer(estimator=DecisionTreeRegressor())
X_titamic_train_complete = imputer.fit_transform(X_titamic_train_encoded)
# Đầu ra là 'numpy.ndarray' vì vậy chúng ta chuyển nó và dataframe để nhất quán
X_titamic_train_complete = pd.DataFrame(X_titamic_train_complete, columns=X_titamic_train_encoded.columns)
print('Số lượng giá trị thiếu: \n', X_titamic_train_complete.isnull().sum())
```

## Áp dụng mô hình đã học và dữ liệu thử nghiệm chưa hoàn thiện

đầu tiên, áp dụng chuẩn hóa

```python
# Bộ mã hóa_sex đã học phải được sử dụng để mã hóa dữ liệu thử nghiệm, LƯU Ý KHÔNG có sự phù hợp ở đây, chỉ cần chuyển đổi
X_titanic_test_encoded = X_titanic_test.copy()
X_titanic_test_encoded['sex'] = encoder_sex.transform(X_titanic_test_encoded['sex'].values.reshape(-1,1))

#Bộ mã hóa đã học2 nên được sử dụng để mã hóa dữ liệu thử nghiệm, LƯU Ý KHÔNG có sự phù hợp ở đây, chỉ cần chuyển đổi
X_titanic_test_encoded['cabin'] = encoder_cabin.transform(X_titanic_test_encoded['cabin'].values.reshape(-1,1).astype(str))

# Bây giờ, lấy lại các giá trị nan bị thiếu trong dữ liệu được mã hóa
X_titanic_test_encoded['cabin'].replace(cabin_nan_code, np.nan, inplace = True)
```

Thứ hai, sử dụng bộ nhập đã học để ước tính các giá trị bị thiếu trong dữ liệu thử nghiệm


```python
print('Số lượng giá trị thiếu trong tập dữ liệu thử nghiệm trước khi bù khuyết: \n', X_titanic_test_encoded.isnull().sum())
X_titanic_test_complete = imputer.transform(X_titanic_test_encoded)
X_titanic_test_complete = pd.DataFrame(X_titanic_test_complete, columns = X_titanic_test_encoded.columns)
print('Số lượng giá trị thiếu trong tập dữ liệu thử nghiệp sau khi bù khuyết: \n', X_titanic_test_complete.isnull().sum())
```

Cuối cùng, chúng ta biểu diễn phân loại bằng dữ liệu bù khuyết đã hoàn thiện

```python
# Chúng ta sử dụng f-measure vì những lớp thì không balanced
from sklearn.metrics import f1_score
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier(random_state=0)
#classifier = SVC()
classifier.fit(X_titamic_train_complete, y_titanic_train)
print('F1 score sau bù khuyết = ', f1_score(classifier.predict(X_titanic_test_complete), y_titanic_test))
```
